```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Sample - Superstore.csv", encoding = 'latin-1') 
```

```python
data.head()
```

```python
data.describe()
```

```python
data.info()
```

# Converting date columns

```python
data['Order Date'] = pd.to_datetime(data['Order Date'])
data['Ship Date'] = pd.to_datetime(data['Ship Date'])
```

```python
data.info()
```

```python
data['Order Year'] = data['Order Date'].dt.year
data['Order Month'] = data['Order Date'].dt.month
data['Order Day Of Week'] = data['Order Date'].dt.dayofweek
```

```python
data.head()
```

# Monthly sales analysis

```python
sales_by_month = data.groupby('Order Month')['Sales'].sum().reset_index()
```

```python
sales_by_month
```

```python
plt.figure(figsize=(8,5))
plt.plot(sales_by_month.index, sales_by_month.values, marker='o')

plt.title("sales_by_month")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

```python
data.head()
```

# Sales by category

```python
sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()
```

```python
sales_by_category
```

```python
plt.figure(figsize=(6,6))

plt.pie(sales_by_category['Sales'], 
        labels=sales_by_category['Category'], 
        autopct='%1.1f%%',
        startangle=90)
colors = plt.cm.Pastel1.colors
plt.title("Sales by Category")

plt.show()
```

```python
data.head()
```

# Sales analysis by sub category

```python
sales_by_subcategory = data.groupby('Sub-Category')['Sales'].sum().reset_index()
```

```python
sales_by_subcategory
```

```python
plt.figure(figsize=(10,6))

plt.bar(sales_by_subcategory['Sub-Category'],
        sales_by_subcategory['Sales'])

plt.title("Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.show()
```

# Sales And Profit By Customer Segment

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales':'sum' , 'Profit':'sum'}).reset_index()
```

```python
sales_profit_by_segment
```

```python
sales_profit_by_segment.set_index('Segment').plot(kind='bar', figsize=(8,5))

plt.title("Sales and Profit by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Amount")

plt.xticks(rotation=0)

plt.show()
```

# Sales To Profit Ratio

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales':'sum' , 'Profit':'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales'] / sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment', 'Sales_to_Profit_Ratio']])
```
